```python
import pandas as pd
import glob
path = "experiment_results/data/random/"  # Replace with your folder path
csv_files = glob.glob(path + '/*.csv')
all_data = []

for filename in csv_files:
    df = pd.read_csv(filename)
    all_data.append(df)

random_results = pd.concat(all_data, ignore_index=True)
random_results["experiment"] = "random"
random_results
```
