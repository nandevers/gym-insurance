# Article Analytics Notebook
This notebook contains an analysis of article performance data. It reads data from multiple parquet files, processes them using pandas, and performs various aggregation and pivot table operations. Please ensure that all necessary data files are available in the specified directories.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# Add more imports as necessary
```

## Data Loading
We will attempt to load the necessary parquet files. If any files are missing, an error will be raised.

```
try:
    rdata_1 = pd.read_parquet('./temp/data/experiment_results/random/insurenv_results_500000.parquet').assign(budget=500000)
    rdata_2 = pd.read_parquet('./temp/data/experiment_results/random/insurenv_results_1000000.parquet').assign(budget=1000000)
    rdata_3 = pd.read_parquet('./temp/data/experiment_results/random/insurenv_results_10000000.parquet').assign(budget=10000000)
except FileNotFoundError as e:
    print(f'Error loading data: {e}')
```

## Data Processing
Here we concatenate the loaded datasets and perform further data processing.

```
if 'rdata_1' in locals() and 'rdata_2' in locals() and 'rdata_3' in locals():
    full_data = pd.concat([rdata_1, rdata_2, rdata_3])
    print(full_data.head())
else:
    print('Data could not be processed due to missing files.')
```

## Data Analysis
We perform a pivot table operation to analyze the article performance by budget and decision metrics.

```
if 'full_data' in locals():
    pivot_result = full_data.pivot_table(index=['experiment', 'initial_budget', 'episodes'], columns='decision', values='wc', aggfunc='count')
    print(pivot_result)
else:
    print('Data not available for pivot table analysis.')
```

## Conclusion
In this notebook, we loaded article data from multiple parquet files, processed the data, and analyzed it using a pivot table to explore trends across different budgets and decisions.
