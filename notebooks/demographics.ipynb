```python
import pandas as pd 

import pandas as pd

# Define the file path to the Parquet file
file_path = "../.temp/data/processed/psr_full_history.parquet"

# Read data from the Parquet file
data = pd.read_parquet(file_path)
```

```python
geographical_level = 'sg_uf_propriedade'
```

```python
data['dt_proposta'] = pd.to_datetime(data['dt_proposta'], format="%d%m%Y")

data['year_proposta'] = data['dt_proposta'].dt.year
```

```python
import pandas as pd
import plotly.express as px

# Sample data loading (replace 'data' with your actual DataFrame)
# data = pd.read_csv('your_data.csv')

# Group by state and year, count unique proposals
state_year_claims = data.groupby([geographical_level, 'year_proposta'])['nr_proposta'].nunique().reset_index()

# Sum total claims by state
state_totals = state_year_claims.groupby(geographical_level)['nr_proposta'].sum().reset_index()

n = 5
# Get the top N states and group others as 'Other'
top_states = state_totals.nlargest(n, 'nr_proposta')[geographical_level].tolist()
state_year_claims[geographical_level] = state_year_claims[geographical_level].apply(lambda x: x if x in top_states else 'Other')
# Aggregate 'Other' states
state_year_claims = state_year_claims.groupby([geographical_level, 'year_proposta'])['nr_proposta'].sum().reset_index()

# Calculate total number of claims per year
year_totals = state_year_claims.groupby('year_proposta')['nr_proposta'].sum().reset_index()

# Merge total claims to calculate percentages
state_year_claims = pd.merge(state_year_claims, year_totals, on='year_proposta', suffixes=('', '_total'))

# Calculate percentage contribution for each state
state_year_claims['percentage'] = state_year_claims['nr_proposta'] / state_year_claims['nr_proposta_total'] * 100
state_year_claims['percentage'] = state_year_claims['percentage'].round()

# Create a 100% stacked bar chart (relative proportions) with absolute values on top
fig = px.bar(
    state_year_claims, 
    x='year_proposta', 
    y='percentage', 
    color=geographical_level, 
    title=f'Claims Distribution Over Time (Top {n} States vs Others)',
    labels={'year_proposta': 'Year', 'percentage': 'Percentage of Total Claims', geographical_level: 'State'},
    text_auto=True  # Auto display the percentages inside the bars
)

# Add absolute total claims as text on top of the bars
for year, total in zip(year_totals['year_proposta'], year_totals['nr_proposta']):
    fig.add_annotation(
        x=year,
        y=103,  # Position the label at the top of each bar (100%)
        text=f"{total:,}",  # Format the total with commas
        showarrow=False,
        yshift=10,  # Shift the text slightly above the bar
        font=dict(size=12),
    )

# Update layout to format the chart
fig.update_layout(
    height=700,
    width=1500, 
    barmode='stack',  # Stack bars to 100%
    xaxis_title='Year',
    yaxis_title='Percentage of Total Claims',
    yaxis=dict(ticksuffix='%'),  # Show y-axis as percentages
    legend_title='State'
)

# Show the plot
fig.show()
```

```python
import pandas as pd
import plotly.express as px

data_pr = pd.read_parquet("C:\\Users\\nande\\Repos\\gym-insurance\\.temp\\data\\processed\\psr_full_history.parquet").query("sg_uf_propriedade=='PR'")
data_pr.columns
```

```python
# Convert the date column to datetime if not already in datetime format
data_pr['dt_proposta'] = pd.to_datetime(data_pr['dt_proposta'])
# Identify the top 10 most frequent 'nm_cultura_global' values
top_10_culturas = data_pr['nm_cultura_global'].value_counts().nlargest(5).index

# Replace all other 'nm_cultura_global' values with 'Others'
data_pr['nm_cultura_global_top10'] = data_pr['nm_cultura_global'].where(data_pr['nm_cultura_global'].isin(top_10_culturas), 'Others')

# Group by 'dt_proposta' (time) and 'nm_cultura_global_top10'
grouped_data = data_pr.groupby([data_pr['dt_proposta'].dt.to_period('Y'), 'nm_cultura_global_top10']).size().unstack(fill_value=0)

# Reset index to get a standard time series index
grouped_data.index = grouped_data.index.to_timestamp()

# Normalize to get percentages (100% stacked)
grouped_data_percentage = grouped_data.div(grouped_data.sum(axis=1), axis=0) * 100

# Convert the normalized dataframe back to a long format for Plotly
df_long = grouped_data_percentage.reset_index().melt(id_vars='dt_proposta', var_name='nm_cultura_global_top10', value_name='percentage')

# Create the 100% stacked bar chart using Plotly
fig = px.bar(df_long, 
             x='dt_proposta', 
             y='percentage', 
             color='nm_cultura_global_top10',
             labels={'percentage': 'Percentage', 'dt_proposta': 'Time'},
             title='100% Stacked Bar Chart of Top  5 nm_cultura_global Over Time (by dt_proposta)')

# Update layout for better readability
fig.update_layout(barmode='stack', xaxis={'categoryorder': 'category ascending'}, 
                  xaxis_title='Time', yaxis_title='Percentage')

# Show the plot
fig.show()
```

```python
import plotly.express as px
import pandas as pd

# Sample data for demonstration
data = {
    'dt_proposta': pd.date_range(start='2018-01-01', periods=100, freq='M'),
    'nm_cultura_global': ['Culture ' + str(i % 7) for i in range(100)]  # Some repeating cultures
}
data_pr = pd.DataFrame(data)

# Convert 'dt_proposta' to datetime if needed
data_pr['dt_proposta'] = pd.to_datetime(data_pr['dt_proposta'])

# Identify the top 5 most frequent 'nm_cultura_global' values
top_5_culturas = data_pr['nm_cultura_global'].value_counts().nlargest(5).index

# Label 'Others' for all values not in the top 5
data_pr['nm_cultura_global_top5'] = data_pr['nm_cultura_global'].where(data_pr['nm_cultura_global'].isin(top_5_culturas), 'Others')

# Group by 'dt_proposta' (year) and 'nm_cultura_global_top5' for time-series analysis
grouped_data = data_pr.groupby([data_pr['dt_proposta'].dt.to_period('Y'), 'nm_cultura_global_top5']).size().unstack(fill_value=0)

# Convert index to timestamp and normalize for percentages (100% stacked bar chart)
grouped_data.index = grouped_data.index.to_timestamp()
grouped_data_percentage = grouped_data.div(grouped_data.sum(axis=1), axis=0) * 100

# Transform data to long format for Plotly
df_long = grouped_data_percentage.reset_index().melt(id_vars='dt_proposta', var_name='nm_cultura_global_top5', value_name='Percentage')

# Create the 100% stacked bar chart
fig = px.bar(df_long, 
             x='dt_proposta', 
             y='Percentage', 
             color='nm_cultura_global_top5',
             labels={'Percentage': 'Percentage of Total', 'dt_proposta': 'Year'},
             title='Top 5 Cultures Over Time with Proportion of Total',
             color_discrete_map={'Others': 'lightgrey'})

# Update layout for readability and focus
fig.update_layout(barmode='stack', 
                  xaxis=dict(title='Year', tickformat='%Y'), 
                  yaxis=dict(title='Percentage of Total (%)'), 
                  title_x=0.5,
                  legend_title_text='Culture Type')

# Customize hover information for better context
fig.update_traces(hovertemplate='<b>%{x|%Y}</b><br>Category: %{color}<br>Percentage: %{y:.1f}%')

# Save the plot
file_path_html = 'top_5_cultures_over_time.html'
file_path_png = 'top_5_cultures_over_time.png'

# Save as HTML for interactivity
#fig.write_html(file_path_html)

# Save as PNG for static viewing
fig.write_image(file_path_png)

fig.show()

```
