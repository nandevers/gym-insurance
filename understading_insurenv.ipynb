# Understanding InsurEnv

In this notebook, we will delve into the `InsurEnv` environment, focusing on understanding how it works step by step, and particularly analyzing the rewards function.


```python
import pandas as pd
import numpy as np
from gym_insurance.envs.insurenv import InsurEnv
```

```python
# Create a sample dataset
data = pd.DataFrame({
    'index_column': range(10),
    'value_column': np.random.rand(10) * 1000,
    'feature_1': np.random.rand(10),
    'feature_2': np.random.rand(10)
})

# Initialize the InsurEnv environment
env = InsurEnv(
    data=data,
    index_column='index_column',
    value_column='value_column',
    state_columns=['feature_1', 'feature_2'],
    budget=100000
)
```

## Environment Initialization

When the `InsurEnv` environment is initialized, it takes in a dataset, column names for indexing, value, and state features, as well as an initial budget. The environment is now ready to be interacted with, and it's in its initial state.
