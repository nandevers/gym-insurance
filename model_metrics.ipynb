```python
import pandas as pd
import seaborn as sns
import plotly.express as px
import os
filename = 'experiment_results/results_1000000.paquet'

# Function to load data
def load_data(filename):
    return pd.read_parquet(filename)
```

```python
def analyze_data(df):
    analytics = {}
    analytics['total_rewards'] = df['rewards'].sum()
    analytics['average_rewards'] = df['rewards'].mean()
    analytics['max_reward'] = df['rewards'].max()
    analytics['cumulative_rewards_over_time'] = df['cum_rewards'].iloc[-1]
    analytics['average_cumulative_rewards'] = df['cum_rewards'].mean()
    analytics['budget_utilization'] = (df['initial_budget'] - df['budget']).mean()
    analytics['average_steps'] = df['steps'].mean()
    analytics['success_rate'] = (df['rewards'] > 0).mean()  # Define success as a positive reward
    analytics['average_value_of_indemnization'] = df['qt_valor_indenizacao'].mean()
    analytics['std_dev_rewards'] = df['rewards'].std()
    return analytics

# Function to generate plots
def generate_plots(df):
    os.makedirs('results/plots', exist_ok=True)
    
    # Example plot with seaborn
    sns_plot = sns.lineplot(data=df, x='steps', y='cum_rewards')
    sns_plot.figure.savefig('results/plots/cum_rewards_plot.png')
    
    # Example plot with Plotly
    fig = px.line(df, x='steps', y='rewards', title='Rewards over Steps')
    fig.write_image('results/plots/rewards_plot.png')
```
